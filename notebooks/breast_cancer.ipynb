```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
from sklearn.preprocessing import StandardScaler                                                                              
```

```python
# train_data_path = "../data/logs/run_0/breast_train/synth_aim_l1_eps_1.0.csv"
train_data_path = "../data/diabetes_train.csv"
test_path = "../data/diabetes_test.csv"
```

```python
def train(model, X_train, y_train):
    model.fit(X_train, y_train)
    return model
```

```python
def test(model, X_test, y_test):
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    return accuracy, f1
```

```python
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_path)
```

```python
X_train = train_df.drop(columns=["class"])
y_train = train_df["class"]
X_test = test_df.drop(columns=["class"])
y_test = test_df["class"]
```

```python
model = LogisticRegression(max_iter=1000)
model = train(model, X_train, y_train)
```

```python
accuracy, f1 = test(model, X_test, y_test)
```

```python
print(f"Accuracy: {accuracy:.4f}")
print(f"F1 Score: {f1:.4f}")
```

```python
def run(train_path, test_path):
    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)

    X_train = train_df.drop(columns=["class"])
    
    y_train = train_df["class"]
    X_test = test_df.drop(columns=["class"]).astype("float64")
    y_test = test_df["class"]                                                                                      

    model = LogisticRegression(max_iter=1000, random_state=42)
    model = train(model, X_train, y_train)
    accuracy, f1 = test(model, X_test, y_test)
    return accuracy, f1
```

## Real Data

```python
train_data_path = "../data/breast_train.csv"
test_path = "../data/breast_test.csv"
accuracy, f1 = run(train_data_path, test_path)
print(f"Accuracy: {accuracy:.4f}")
print(f"F1 Score: {f1:.4f}")
```
