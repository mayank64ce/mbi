# Diabetes dataset

```python
diabetes_path = "../data/diabetes.csv"
```

```python
db = pd.read_csv(diabetes_path)
```

```python
db
```

```python
db.value_counts()
```

```python
db.columns
```

```python
for c in ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age']:
    print(c)
    db[c] = pd.qcut(db[c].values, 5, duplicates = 'drop').codes + 1

    print('************')
```

```python
domain_db = {}
for c in db.columns:
    print(c)
    print(len(db[c].value_counts()))
    domain_db[c] = len(db[c].value_counts())
    print('************')
```

```python
domain_db
```

```python
len(db)
```

```python
train_db = db.sample(n=(round(0.8*len(db))))
test_db = db[~db.index.isin(train_db.index)]
```

```python
train_db.to_csv('diabetes_train.csv', index=False)
test_db.to_csv('diabetes_test.csv', index=False)
```
